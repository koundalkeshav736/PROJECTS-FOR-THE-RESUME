```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("zomato (1).csv", encoding="latin1")
print(df)
```

```python
print(df.duplicated().sum())
```

```python
print(df.drop_duplicates(inplace=True))
```

```python
print(df.size)
```

```python
plt.figure(figsize=(12,6))
plt.show()
sns.countplot(y='City',data=df,order=df['City'].value_counts().head(15).index)
plt.title("top 15 CITIES by no of restaurants")
plt.xlabel("no of restaurants")
plt.ylabel("city")
plt.show()
```

```python
#TOP 10 cuisine
top=df['Cuisines'].value_counts().head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top.values,y=top.index)
plt.show()
```

```python
#ratings and comments
plt.figure(figsize=(10,5))
sns.histplot(df["Aggregate rating"],bins=20,kde=True)
plt.show()
```

```python
#ratings and comments
plt.figure(figsize=(10,5))
sns.boxplot(x='Aggregate rating',y='Rating color',data=df)
plt.show()
#
```

```python
sns.countplot(x='Has Online delivery',data=df)
```

```python
sns.countplot(x='Has Table booking',data=df)
plt.show()
```

```python
delivery=df['Has Online delivery'].value_counts()
plt.pie(delivery,labels=delivery.index,autopct='%1.2f%%')
plt.show()
```

```python
sns.countplot('Price Range',data=df)
plt.show()
```

```python
sns.hitplot(df['Average Cost for two'],bins=10,kde)
```

```python
top=df.sort_values('Aggregate rating',ascending=False).head(10)
sns.barplot(data=top,x='Aggregate rating',y='Restaurant Name')
plt.show()
```

```python
top=df.sort_values('Votes',ascending=False).head(10)
sns.barplot(data=top,x='Votes',y='Restaurant Name')
plt.show()
```

```python
sns.countplot(y='Rating text',data=df,order=df['Rating text'].value_counts().index)
plt.show()
```

```python
rating=df.groupby('Price range')['Aggregate rating'].mean()
rating.plot(kind='bar',figsize=(8,5))
plt.title("Average Aggregate Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Aggregate Rating")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True,cmap='coolwarm')
plt.show()
```

```python
#pair ploting of the data set
sns.pairplot(df[['Average Cost for two', 'Aggregate rating', 'Price range']])
plt.show()
```

```python
sns.scatterplot(y='Aggregate rating',x='Average Cost for two',data=df)
```

```python
city=df.groupby("City")['Average Cost for two'].mean().sort_values(ascending=False).head(10)
city.plot(kind='bar',color='green')
plt.show()
```

```python
booking=df.groupby('Has Table booking')['Aggregate rating'].mean()
print(booking)
```

```python
sns.boxplot(x='Has Online delivery',y='Aggregate rating',data=df)
plt.show()
```

```python
sns.boxplot(x='Price range',y='Aggregate rating',data=df)
plt.show()
```
