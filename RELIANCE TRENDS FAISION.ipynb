```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rt=pd.read_csv('Reliance Trends Fashion.csv',encoding='Latin1')
print(rt)
```

```python
print(rt.duplicated().sum())
```

```python
print(rt.drop_duplicates(inplace=True))
```

```python
print(rt.size)
```

```python
print(rt.isnull().sum())
```

```python
print(rt.head(15))
```

```python
rt['Discount_Price (in Rs.)'] = rt['Discount_Price (in Rs.)'].str.replace(',', '').astype(float)
```

```python
print(rt['Discount_Price (in Rs.)'].head())
```

```python
print(rt.columns)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=rt,x='Brand',y='Category',hue=True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=rt,x='Discount_Price (in Rs.)',bins=20,kde=True)
plt.show()
```

```python
small=rt.sort_values('Brand',ascending=True).head(15)
plt.figure(figsize=(10,5))
sns.histplot(data=small,x='Brand',y='Category',hue=True,bins=20)
plt.show()
```

```python
gender=rt['Category_by_gender'].value_counts()
plt.pie(gender,labels=gender.index,autopct='%1.2f%%')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=rt,x='Category',y='Category_by_gender',hue=True)
plt.show()
```

```python
rt['Original_Price (in Rs.)'] = rt['Original_Price (in Rs.)'].str.replace(',', '').astype(float)
print(rt['Original_Price (in Rs.)'].head(10))
```

```python
rt['change']=rt['Original_Price (in Rs.)']-rt['Discount_Price (in Rs.)']
print(rt['change'].head(10))
```

```python

plt.figure(figsize=(10,5))
sns.histplot(rt['change'], bins=20, kde=True)
plt.show()
```

```python
print(rt.columns)
```

```python
top=rt.sort_values('Brand',ascending=True).head(15)
plt.figure(figsize=(10,5))
sns.pairplot(data=top, vars=['Category_by_gender','Brand','Category'])
plt.show()
```
