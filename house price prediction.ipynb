```python
%%writefile app.py
import streamlit as st
import pandas as pd
from sklearn.linear_model import LinearRegression

# Page configuration
st.set_page_config(page_title="Housing Price Predictor", layout="centered")

# Title and Description
st.title("🏡 Housing Price Prediction App")
st.write("This app predicts house prices based on features from your uploaded dataset.")

# Load dataset
try:
    df = pd.read_csv('extracted_data/Housing.csv')
except Exception:
    # Fallback to local Housing.csv if path differs
    df = pd.read_csv('Housing.csv')

# Features and target variable
features = ['area', 'bedrooms', 'bathrooms', 'stories', 'parking']
X = df[features]
y = df['price']

# Train Linear Regression model
model = LinearRegression()
model.fit(X, y)

# User Inputs
st.subheader("Enter House Specifications:")
col1, col2 = st.columns(2)

with col1:
    area = st.number_input("Area (sqft)", min_value=int(X['area'].min()), max_value=int(X['area'].max()), value=5000, step=100)
    bedrooms = st.slider("Number of Bedrooms", min_value=int(X['bedrooms'].min()), max_value=int(X['bedrooms'].max()), value=3)
    bathrooms = st.slider("Number of Bathrooms", min_value=int(X['bathrooms'].min()), max_value=int(X['bathrooms'].max()), value=1)

with col2:
    stories = st.slider("Number of Stories", min_value=int(X['stories'].min()), max_value=int(X['stories'].max()), value=1)
    parking = st.slider("Parking Spaces", min_value=int(X['parking'].min()), max_value=int(X['parking'].max()), value=1)

# Predict button
if st.button("Predict Price"):
    input_data = [[area, bedrooms, bathrooms, stories, parking]]
    prediction = model.predict(input_data)[0]
    st.success(f"💰 Predicted House Price: ${prediction:,.2f}")
```
